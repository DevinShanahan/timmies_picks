```python
import itertools

import numpy as np
import gym
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
class WordleEnv(gym.Env):
    def __init__(self):
        with open('/Users/devin/Downloads/wordle_allowed_words.txt', 'r', encoding='utf-8') as f:
            self.allowed = [w.strip() for w in f.readlines()]
        
        with open('/Users/devin/Downloads/wordle_possible_solutions.txt', 'r', encoding='utf-8') as f:
            self.solutions = [w.strip() for w in f.readlines()]
            
        alphabet = 'abcdefghijklmnopqrstuvwxyz'
        self.alpha = {l: i for i, l in enumerate(alphabet)}
        self.ints = {i: l for i, l in enumerate(alphabet)}
        self.obs_map = {tup: i + 1 for i, tup in enumerate(itertools.product(*[[0, 1, 2]] * 5))}
        self.action_space = gym.spaces.Discrete(len(self.allowed))
        self.observation_space = gym.spaces.Discrete(3**5 + 1)
        self.solution = self.solutions[np.random.randint(0, len(self.solutions))]
        self.tries = 0
        self.lastword = None
        self.choices = []
        
    def encode(self, word):
        return (self.alpha[l] for l in word)
        
    def decode(self, encoded):
        return ''.join([self.ints[i] for i in encoded])
    
    def step(self, action):
        assert action in self.action_space
        self.lastword = self.allowed[action]
        self.choices.append(self.lastword)
        self.tries += 1
        
        observation = []
        reward = 0
        for a_l, s_l in zip(self.lastword, self.solution):
            if a_l == s_l:
                observation.append(2)
                reward += 10
            elif a_l in self.solution:
                observation.append(1)
                reward += 3
            else:
                observation.append(0)
                reward -= 2
        if reward == 50:
            reward *= 10 * (7 - self.tries)
            done = True
#             print('correct', self.tries)
        elif self.tries < 6:
            done = False
        else:
            done = True
        return self.obs_map[tuple(observation)], reward, done, {}

    def reset(self, *, seed=None, options=None):
        self.solution = self.solutions[np.random.randint(0, len(self.solutions))]
        self.tries = 0
        self.choices = []
        return 0
    
    def render(self):
        print(self.choices)
```
